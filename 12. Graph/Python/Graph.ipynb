# **Adjacency List (Unweighted and Directed)**

image.png

```python
class Node:

  def __init__(self, destination):
    self.destination = destination
    self.next = None
```

```python
n = 4 #No of Nodes/Vertices
```

```python
adjList = np.array([None]*(n + 1)) #Creating an empty adjacency list(array) of n length
```

```python
def addEdgetoList(source, destination):
    edge = Node(destination) #Creating a node object
    if adjList[source] == None:
        adjList[source] = edge
    else:
        temp = adjList[source]
        while temp.next != None:
            temp = temp.next
        temp.next = edge
```

```python
addEdgetoList(1, 2)
addEdgetoList(2, 4)
addEdgetoList(3, 1)
addEdgetoList(3, 4)
addEdgetoList(3, 2)
```

```python
def printAdjacencyList(adjList):
    for i in range(len(adjList)):
        temp = adjList[i]
        print(f"{i}: ", end = " ")
        while temp != None:
            print(temp.destination, end = "->")
            temp = temp.next
        print()
```

```python
printAdjacencyList(adjList)
```

# **Adjacency List (Weighted and Undirected)**

```python
class Node:

  def __init__(self, destination, weight):
    self.destination = destination
    self.weight = weight
    self.next = None
```

```python
n = 4 #No of Nodes/Vertices
```

```python
adjList = np.array([None]*(n + 1)) #Creating an empty adjacency list(array) of n length
```

```python
def addEdgetoList(source, destination, weight):
    edge1 = Node(destination, weight)
    edge2 = Node(source, weight)

    if adjList[source] == None:
        adjList[source] = edge1
    else:
        temp = adjList[source]
        while temp.next != None:
            temp = temp.next
        temp.next = edge1


    if adjList[destination] == None:
      adjList[destination] = edge2
    else:
        temp = adjList[destination]
        while temp.next != None:
            temp = temp.next
        temp.next = edge2
```

```python
addEdgetoList(1, 2, 6)
addEdgetoList(2, 4, 3)
addEdgetoList(3, 1, 7)
addEdgetoList(3, 4, 8)
addEdgetoList(3, 2, 1)
```

```python
def printAdjacencyList(adjList):
    for i in range(len(adjList)):
        temp = adjList[i]
        print(f"{i}: ", end = " ")
        while temp != None:
            print(f'{temp.destination} ({temp.weight})', end = "->")
            temp = temp.next
        print()
```

```python
printAdjacencyList(adjList)
```

# **Adjacency Matrix (Weighted and Undirected)**

```python
adjMat = np.zeros((n + 1, n + 1), dtype = int)
```

```python
def addEdgetoMatrix(source, destination, weight):
  adjMat[source][destination] = weight
  adjMat[destination][source] = weight
```

```python
addEdgetoMatrix(1, 2, 6)
addEdgetoMatrix(2, 4, 3)
addEdgetoMatrix(3, 1, 7)
addEdgetoMatrix(3, 4, 8)
addEdgetoMatrix(3, 2, 1)
addEdgetoMatrix(4, 3, 9)
```
